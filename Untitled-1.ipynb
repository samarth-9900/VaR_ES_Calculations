### Portfolio and Risk Factors

The portfolio consists of different types of financial instruments/securities such as equities, equity indices, commodities, bonds, and foreign exchange.

Each security is associated with one or more risk factors that can affect its value. A single security may depend on multiple risk factors, while the same risk factor can affect multiple securities.

For this project:

- **Equities** → stock price
- **Equity index** → index level
- **Commodity** → commodity price
- **Bonds** → Treasury yield and inflation
- **Forex** → exchange rate

The portfolio weights represent the proportion of total portfolio capital invested in each security.

### Import Historical Data

````python
# ============================================================
# IMPORT EQUITY RISK-FACTOR DATA
# ============================================================


# Date range
start_date = "2010-01-01"
end_date = "2026-01-01"


# Equity risk factors
equity_tickers = {
    "AAPL_PRICE": "AAPL",
    "MSFT_PRICE": "MSFT"
}


# Download adjusted closing prices
equity_prices = yf.download(
    list(equity_tickers.values()),
    start=start_date,
    end=end_date,
    auto_adjust=True
)["Close"]


# Rename tickers to our risk-factor names
equity_prices = equity_prices.rename(
    columns={
        "AAPL": "AAPL_PRICE",
        "MSFT": "MSFT_PRICE"
    }
)


# Data inspection
print(equity_prices.info())
````

````python
# ============================================================
# IMPORT EQUITY INDEX RISK-FACTOR DATA
# ============================================================

# Equity index risk factor
index_tickers = {
    "SP500_INDEX": "^GSPC"
}


# Download adjusted closing prices
index_prices = yf.download(
    list(index_tickers.values()),
    start=start_date,
    end=end_date,
    auto_adjust=True
)["Close"]


# Rename ticker to our risk-factor name
index_prices = index_prices.rename(
    columns={
        "^GSPC": "SP500_INDEX"
    }
)


# Data inspection
print(index_prices.info())
````

````python
# ============================================================
# IMPORT COMMODITY RISK-FACTOR DATA
# ============================================================

# Commodity risk factor
commodity_tickers = {
    "GOLD_PRICE": "GC=F"
}


# Download adjusted closing prices
commodity_prices = yf.download(
    list(commodity_tickers.values()),
    start=start_date,
    end=end_date,
    auto_adjust=True
)["Close"]


# Rename ticker to our risk-factor name
commodity_prices = commodity_prices.rename(
    columns={
        "GC=F": "GOLD_PRICE"
    }
)


# Data inspection
print(commodity_prices.info())
````

````python
# ============================================================
# IMPORT BOND(FIXED-Income) RISK FACTORS
# ============================================================

# We import Treasury yield data for different maturities,
# not the price/history of individual bonds.
# Each yield represents a maturity point on the Treasury yield curve.


# ------------------------------------------------------------
# FRED risk-factor series
# ------------------------------------------------------------

fred_factors = {
    "US2Y_YIELD": "DGS2",       # 2-Year Treasury Yield
    "US10Y_YIELD": "DGS10",    # 10-Year Treasury Yield
    "INFLATION": "CPIAUCSL"    # CPI
}


# ------------------------------------------------------------
# Download data from FRED
# ------------------------------------------------------------

bond_risk_factors = web.DataReader(
    list(fred_factors.values()),
    "fred",
    start_date,
    end_date
)


# ------------------------------------------------------------
# Rename FRED series to our risk-factor names
# ------------------------------------------------------------

bond_risk_factors = bond_risk_factors.rename(
    columns={
        "DGS2": "US2Y_YIELD",
        "DGS10": "US10Y_YIELD",
        "CPIAUCSL": "INFLATION"
    }
)

# ============================================================
# EDA - INTERPOLATE MISSING BOND RISK FACTORS(Time-Series Interpolation;Almost Same as Linear Interpolation(method='linear'))
# ============================================================

bond_risk_factors["US2Y_YIELD"] = (
    bond_risk_factors["US2Y_YIELD"]
    .interpolate(method="time", limit_area="inside")
)

bond_risk_factors["US10Y_YIELD"] = (
    bond_risk_factors["US10Y_YIELD"]
    .interpolate(method="time", limit_area="inside")
)

bond_risk_factors["INFLATION"] = (
    bond_risk_factors["INFLATION"]
    .interpolate(method="time", limit_area="inside")
)

# Data inspection
print(bond_risk_factors.info())
````

````python
# ============================================================
# IMPORT FOREX RISK-FACTOR DATA
# ============================================================

# Forex risk factor
forex_tickers = {
    "EURUSD": "EURUSD=X"
}


# Download adjusted closing prices
forex_prices = yf.download(
    list(forex_tickers.values()),
    start=start_date,
    end=end_date,
    auto_adjust=True
)["Close"]


# Rename ticker to our risk-factor name
forex_prices = forex_prices.rename(
    columns={
        "EURUSD=X": "EURUSD"
    }
)


# Data inspection
print(forex_prices.info())
````

### Risk Factor Matrix

````python
# ============================================================
# COMBINE RISK-FACTOR DATA
# ============================================================

risk_factor_data = pd.concat(
    [
        equity_prices,
        index_prices,
        commodity_prices,
        forex_prices,
        bond_risk_factors
    ],
    axis=1,
    join="inner"
)

# Data inspection
print(risk_factor_data.info())
````

````python
# ============================================================
# CREATE RISK-FACTOR MATRIX
# ============================================================

risk_factor_matrix = pd.DataFrame(index=risk_factor_data.index)

# Price risk factors → percentage change
risk_factor_matrix["AAPL_RETURN"] = (
    risk_factor_data["AAPL_PRICE"].pct_change()
)

risk_factor_matrix["MSFT_RETURN"] = (
    risk_factor_data["MSFT_PRICE"].pct_change()
)

risk_factor_matrix["SP500_RETURN"] = (
    risk_factor_data["SP500_INDEX"].pct_change()
)

risk_factor_matrix["GOLD_RETURN"] = (
    risk_factor_data["GOLD_PRICE"].pct_change()
)

risk_factor_matrix["EURUSD_RETURN"] = (
    risk_factor_data["EURUSD"].pct_change()
)


# Yield risk factors → absolute change
risk_factor_matrix["US2Y_YIELD_CHANGE"] = (
    risk_factor_data["US2Y_YIELD"].diff()
)

risk_factor_matrix["US10Y_YIELD_CHANGE"] = (
    risk_factor_data["US10Y_YIELD"].diff()
)


# Inflation → absolute change
risk_factor_matrix["INFLATION_CHANGE"] = (
    risk_factor_data["INFLATION"].diff()
)


# Remove first row created by pct_change() / diff()
risk_factor_matrix = risk_factor_matrix.dropna()

# Check final matrix
print(risk_factor_matrix.info())
````
